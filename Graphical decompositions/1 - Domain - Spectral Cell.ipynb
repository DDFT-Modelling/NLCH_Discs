```python
# 
k = 0.3

fig, ax = plt.subplots(1)

# Make data.
X_c = np.linspace(-1,1, 100)
Y_c = np.linspace(-1,1, 100)
X, Y = np.meshgrid(X_c, Y_c)
R = np.maximum(abs(X),abs(Y))
Z = np.sin(R)

surf = ax.contourf(X * ( 1 + k * (1-Y**2) ), Y * ( 1 + k * (1-X**2) ), R, levels = 10)

plt.plot( X_c,  1 + k * (1 - X_c**2), c = 'xkcd:wine' )
plt.plot( X_c, -1 - k * (1 - X_c**2), c = 'xkcd:wine' )
plt.plot(  1 + k * (1 - Y_c**2), Y_c, c = 'xkcd:red' )
plt.plot( -1 - k * (1 - Y_c**2), Y_c, c = 'xkcd:red' )

plt.ylim(-1.5,1.5)
plt.xlim(-1.5,1.5)
plt.show()
```
